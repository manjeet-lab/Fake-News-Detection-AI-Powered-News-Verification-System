# 📰 Fake News Detection — Dataset Collection & Basic Data Analysis

This notebook covers the **first stage** of the Fake News Detection project:

1. Loading the raw datasets (`Fake.csv` and `True.csv`)
2. Labeling and combining them into a single dataset
3. Performing basic exploratory checks (shape, columns, missing values, duplicates)
4. Cleaning (removing duplicates), shuffling, and resetting the index
5. Saving the cleaned dataset as `news_dataset.csv`
6. Visualizing the class distribution (Fake vs Real)

> ⚠️ **Note:** This notebook only covers data collection and basic analysis.
> NLP preprocessing, TF-IDF, and Machine Learning model training will be done in later notebooks.


## 1. Import Required Libraries

We start by importing the libraries we need:
- **pandas** → for loading and manipulating the datasets
- **matplotlib / seaborn** → for visualizing the class distribution


```python
# Import necessary libraries
import pandas as pd               # For data loading and manipulation
import matplotlib.pyplot as plt   # For plotting
import seaborn as sns             # For nicer statistical plots

# Make plots look a bit nicer
sns.set_style("whitegrid")
```

## 2. Load the Datasets

We load `Fake.csv` and `True.csv` using Pandas. Both files are assumed to be
in the same folder as this notebook.

Each dataset has the following columns:
- `title` → headline of the article
- `text` → full article body
- `subject` → news category/subject
- `date` → publication date


```python
# Load the Fake and True news datasets
fake_df = pd.read_csv("../data/Fake.csv")
true_df = pd.read_csv("../data/True.csv")

print("Datasets loaded successfully!")
```

## 3. Inspect the Shape and Columns of Both Datasets

Before combining the datasets, let's check:
- How many rows and columns each dataset has
- The column names in each dataset


```python
# Shape (rows, columns) of both datasets
print("Fake news dataset shape:", fake_df.shape)
print("True news dataset shape:", true_df.shape)

# Column names of both datasets
print("\nFake news columns:", fake_df.columns.tolist())
print("True news columns:", true_df.columns.tolist())
```

## 4. Preview the First 5 Rows

Let's take a quick look at the first 5 rows of each dataset to understand
what the data looks like.


```python
# First 5 rows of the fake news dataset
fake_df.head()
```

```python
# First 5 rows of the true news dataset
true_df.head()
```

## 5. Add a `label` Column

To train a machine learning model later, we need a numeric label:
- `Fake.csv` → **label = 0** (Fake news)
- `True.csv` → **label = 1** (Real news)


```python
# Add label column: 0 for fake news, 1 for real news
fake_df["label"] = 0
true_df["label"] = 1

print("Labels added successfully!")
fake_df[["title", "label"]].head(3)
```

## 6. Combine Both Datasets into One DataFrame

Now we concatenate the two labeled datasets into a single DataFrame
using `pd.concat()`, which is efficient and avoids looping row by row.


```python
# Combine both datasets into a single DataFrame
combined_df = pd.concat([fake_df, true_df], axis=0, ignore_index=True)

print("Combined dataset shape:", combined_df.shape)
combined_df.head()
```

## 7. Count of Fake vs Real News Articles

Let's check how many fake and real news articles are present in the
combined dataset. This is an important check for class balance.


```python
# Count number of fake (0) and real (1) news articles
label_counts = combined_df["label"].value_counts()

print("Number of Fake news articles (label=0):", label_counts.get(0, 0))
print("Number of Real news articles (label=1):", label_counts.get(1, 0))
```

## 8. Check for Missing Values

We check each column for missing (null) values in the combined dataset.


```python
# Check missing values in each column
missing_values = combined_df.isnull().sum()
print("Missing values per column:\n")
print(missing_values)
```

## 9. Check for Duplicate Rows

Duplicate rows can bias the model during training, so we first check
how many exact duplicate rows exist in the combined dataset.


```python
# Check number of duplicate rows
duplicate_count = combined_df.duplicated().sum()
print("Number of duplicate rows found:", duplicate_count)
```

## 10. Remove Duplicate Rows

If duplicates were found in the previous step, we remove them here
using `drop_duplicates()`, which operates efficiently on the whole
DataFrame (no manual row-by-row looping).


```python
# Remove duplicate rows (if any)
before_rows = combined_df.shape[0]
combined_df = combined_df.drop_duplicates()
after_rows = combined_df.shape[0]

print(f"Rows before removing duplicates: {before_rows}")
print(f"Rows after removing duplicates: {after_rows}")
print(f"Total duplicate rows removed: {before_rows - after_rows}")
```

## 11. Shuffle the Combined Dataset

Since the combined dataset currently has all fake news rows first,
followed by all real news rows, we shuffle it randomly so that the
labels are well-mixed. This is important before splitting the data
for training/testing in later stages.


```python
# Shuffle the dataset randomly (random_state ensures reproducibility)
combined_df = combined_df.sample(frac=1, random_state=42)

print("Dataset shuffled successfully!")
```

## 12. Reset the Index

After shuffling, the row index is out of order. We reset it so that
the index goes from 0 to n-1 in sequential order.


```python
# Reset the index after shuffling
combined_df = combined_df.reset_index(drop=True)

combined_df.head()
```

## 13. Save the Final Cleaned Dataset

We save the cleaned, shuffled, and re-indexed dataset as `news_dataset.csv`
so it can be used directly in the next stage (NLP preprocessing & modeling).


```python
# Save the final combined dataset to a CSV file
combined_df.to_csv("../data/news_dataset.csv", index=False)

print("Final dataset saved as '../data/news_dataset.csv'")
```

## 14. Visualize Fake vs Real News Distribution

A simple bar chart to visualize the balance between fake and real
news articles in the final dataset.


```python
# Plot the distribution of Fake (0) vs Real (1) news
plt.figure(figsize=(6, 4))
sns.countplot(x="label", data=combined_df, hue="label",
              palette=["#e74c3c", "#2ecc71"], legend=False)
plt.title("Distribution of Fake vs Real News Articles")
plt.xlabel("News Label")
plt.ylabel("Number of Articles")
plt.xticks([0, 1], ["Fake (0)", "Real (1)"])
plt.show()
```

## 15. Final Summary of the Cleaned Dataset

Let's print a final summary covering:
- Total number of articles
- Number of fake vs real articles
- Final shape and columns
- Confirmation that duplicates have been removed


```python
# Print a final summary of the cleaned dataset
print("========== FINAL DATASET SUMMARY ==========")
print(f"Total number of articles     : {combined_df.shape[0]}")
print(f"Total number of columns      : {combined_df.shape[1]}")
print(f"Columns                      : {combined_df.columns.tolist()}")
print(f"Number of Fake articles (0)  : {(combined_df['label'] == 0).sum()}")
print(f"Number of Real articles (1)  : {(combined_df['label'] == 1).sum()}")
print(f"Remaining duplicate rows     : {combined_df.duplicated().sum()}")
print(f"Remaining missing values     : {combined_df.isnull().sum().sum()}")
print("=============================================")
print("\nDataset is now clean, shuffled, and saved as '../data/news_dataset.csv'.")
print("Ready for the next stage: NLP Preprocessing & Feature Engineering.")
```
