## Section 3: Understanding Evaluation Metrics

Before training, let us understand what each evaluation metric means — especially for a fake news classification problem.

### The Confusion Matrix (Foundation of all metrics)

```
                   PREDICTED
              Fake (0)   Real (1)
ACTUAL Fake     TN          FP
       Real     FN          TP
```

| Term | Meaning | Example |
|---|---|---|
| **TP** (True Positive) | Model says Real, Article IS Real | Correct Real classification |
| **TN** (True Negative) | Model says Fake, Article IS Fake | Correct Fake classification |
| **FP** (False Positive) | Model says Real, Article IS Fake | Fake news classified as Real — DANGEROUS |
| **FN** (False Negative) | Model says Fake, Article IS Real | Real news blocked — annoying but less harmful |

---

### The Four Key Metrics

| Metric | Formula | What it means for Fake News |
|---|---|---|
| **Accuracy** | (TP+TN) / Total | Overall % of articles correctly classified |
| **Precision** | TP / (TP+FP) | Of articles predicted Real, how many truly are Real? |
| **Recall** | TP / (TP+FN) | Of actual Real articles, how many did we correctly find? |
| **F1-Score** | 2 x (P x R) / (P+R) | Harmonic mean of Precision and Recall — the balanced measure |
| **ROC-AUC** | Area under ROC curve | Ability to distinguish Fake from Real across all thresholds |

> **Why F1-Score matters most?** In fake news detection, both FP and FN are costly. F1-Score balances Precision and Recall into a single number and is the most reliable metric when both error types matter.

## Section 4: Understanding the Models

### Model 1: Logistic Regression

Despite the name, **Logistic Regression** is a **classification** model, not a regression model.

- It learns a weight for each TF-IDF feature (each word/bigram)
- Positive weight = the word pushes toward **Real News**
- Negative weight = the word pushes toward **Fake News**
- It outputs a **probability** (0.0 to 1.0) via `predict_proba()` — very interpretable!
- Fast to train, works excellently on high-dimensional sparse text data

**Parameters we use:**
- `C=1.0` — regularization strength (controls overfitting)
- `max_iter=1000` — enough iterations to converge
- `solver='lbfgs'` — efficient solver for large datasets
- `random_state=42` — reproducible results

---

### Model 2: Linear Support Vector Machine (LinearSVC)

**LinearSVC** finds the **optimal hyperplane** (decision boundary) that maximally separates Fake News from Real News in the high-dimensional TF-IDF space.

- The 'Support Vectors' are the closest data points to the boundary
- It is the fast, linear version of SVM — ideal for large sparse text matrices
- Typically one of the strongest performers on text classification tasks

> **Important limitation:** `LinearSVC` does **not** implement `predict_proba()`. It is a margin-based classifier, not a probabilistic one — it only exposes `decision_function()`, which returns a signed distance to the separating hyperplane, not a calibrated probability.
>
> We deliberately do **not** wrap it in something like `CalibratedClassifierCV` to fabricate probabilities in this notebook. Instead:
> - We use `decision_function()` scores only where that is statistically valid — e.g. ROC-AUC only needs a score that *ranks* examples correctly, not a true probability, so it remains valid.
> - We do **not** report a "P(Fake)/P(Real)" probability for LinearSVC.
> - Proper, well-calibrated confidence scoring for LinearSVC (via calibration or an alternative model) is deferred to the next task, where it will be handled explicitly.

**Parameters we use:**
- `C=1.0` — controls the margin width (trade-off between margin and misclassification)
- `max_iter=2000` — enough iterations for convergence
- `random_state=42` — reproducible results

## Section 5: Train Model 1 — Logistic Regression

We train Logistic Regression on the TF-IDF training matrix. The model learns which words/bigrams are associated with Fake vs Real news.


````python
# ── Train Logistic Regression ──
print('Training Logistic Regression...')
print('  Input shape :', X_train.shape)
print()

lr_model = LogisticRegression(
    C=1.0,           # Regularization: smaller C = more regularization (prevents overfitting)
    max_iter=1000,   # Maximum number of optimization iterations
    solver='lbfgs',  # Efficient solver for large sparse datasets
    n_jobs=-1,       # Use all available CPU cores
    random_state=42  # Reproducible results
)

start = time.time()
lr_model.fit(X_train, y_train)
lr_train_time = time.time() - start

print('  Logistic Regression trained!')
print('  Training time : {:.2f} seconds'.format(lr_train_time))
print('  Coefficients  : {} (one weight per TF-IDF feature)'.format(len(lr_model.coef_[0])))
````

````python
# ── Generate Logistic Regression predictions ──
# Predicted class labels (0 = Fake, 1 = Real)
lr_pred       = lr_model.predict(X_test)

# Predicted probabilities for each class
# Column 0 = P(Fake), Column 1 = P(Real)
lr_pred_proba = lr_model.predict_proba(X_test)

print('Logistic Regression predictions generated!')
print('  Predictions shape       :', lr_pred.shape)
print('  Probabilities shape     :', lr_pred_proba.shape)
print()
print('Sample predictions (first 10 test articles):')
print('  {:8s}  {:8s}  {:10s}  {:10s}'.format('Actual', 'Predicted', 'P(Fake)', 'P(Real)'))
print('  ' + '-' * 44)
for i in range(10):
    actual = y_test.iloc[i]
    pred   = lr_pred[i]
    p_fake = lr_pred_proba[i][0]
    p_real = lr_pred_proba[i][1]
    status = 'CORRECT' if actual == pred else 'WRONG'
    print('  {:8d}  {:8d}  {:10.4f}  {:10.4f}  {}'.format(actual, pred, p_fake, p_real, status))
````

---

## Section 6: Train Model 2 — Linear Support Vector Machine (LinearSVC)

We use **LinearSVC** — the fast, scalable linear SVM optimized for text classification. It works directly with sparse matrices, making it memory-efficient.

We train it **as-is**, with no probability calibration wrapper. As noted above, this means we will use `decision_function()` (not `predict_proba()`) wherever a continuous score is needed for this model.


````python
# ── Train Linear Support Vector Machine (LinearSVC) ──
print('Training LinearSVC...')
print('  Input shape :', X_train.shape)
print()

svm_model = LinearSVC(
    C=1.0,           # Regularization strength (controls margin width)
    max_iter=2000,   # More iterations for convergence on large sparse data
    random_state=42  # Reproducible results
)

start = time.time()
svm_model.fit(X_train, y_train)
svm_train_time = time.time() - start

print('  LinearSVC trained!')
print('  Training time :', '{:.2f} seconds'.format(svm_train_time))
print('  Coefficients  : {} (one weight per TF-IDF feature)'.format(len(svm_model.coef_[0])))
````

````python
# ── Generate LinearSVC predictions ──
# Predicted class labels (0 = Fake, 1 = Real)
svm_pred = svm_model.predict(X_test)

# LinearSVC has NO predict_proba(). We intentionally do NOT call it here.
# decision_function() gives a signed distance to the hyperplane:
#   > 0  -> classified as Real (1)
#   < 0  -> classified as Fake (0)
#   magnitude -> how far from the boundary (NOT a calibrated probability)
svm_scores = svm_model.decision_function(X_test)

print('LinearSVC predictions generated!')
print('  Predictions shape      :', svm_pred.shape)
print('  Decision scores shape  :', svm_scores.shape)
print('  NOTE: These are raw margin distances, not probabilities.')
print('        LinearSVC does not support predict_proba() out of the box.')
print()
print('Sample predictions (first 10 test articles):')
print('  {:8s}  {:9s}  {:14s}'.format('Actual', 'Predicted', 'Decision Score'))
print('  ' + '-' * 38)
for i in range(10):
    actual = y_test.iloc[i]
    pred   = svm_pred[i]
    score  = svm_scores[i]
    status = 'CORRECT' if actual == pred else 'WRONG'
    print('  {:8d}  {:9d}  {:14.4f}  {}'.format(actual, pred, score, status))
````

## Section 7: Evaluate Both Models

We compute all evaluation metrics for both models and compile them into a comparison table.


````python
# ── Helper function to compute all metrics for a model ──
# The labels are binary (0 = Fake, 1 = Real), so average='binary' with
# pos_label=1 is the statistically correct choice (rather than 'weighted'/'macro').
#
# `y_score` is a continuous score used ONLY for ROC-AUC:
#   - Logistic Regression -> predict_proba()[:, 1]        (a real probability)
#   - LinearSVC            -> decision_function()          (a ranking score, NOT a probability)
# ROC-AUC only requires that the score correctly RANKS examples, so using
# decision_function() here is statistically valid even without calibration.

def evaluate_model(name, y_true, y_pred, y_score, train_time, score_is_probability):
    acc  = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, average='binary', pos_label=1)
    rec  = recall_score(y_true, y_pred, average='binary', pos_label=1)
    f1   = f1_score(y_true, y_pred, average='binary', pos_label=1)
    try:
        auc = roc_auc_score(y_true, y_score)
    except Exception:
        auc = float('nan')
    return {
        'Model'               : name,
        'Accuracy'            : round(acc,  4),
        'Precision'           : round(prec, 4),
        'Recall'              : round(rec,  4),
        'F1 Score'            : round(f1,   4),
        'ROC-AUC'             : round(auc,  4),
        'Training Time'       : round(train_time, 2),
        'Score Type for AUC'  : 'predict_proba' if score_is_probability else 'decision_function (uncalibrated)'
    }

# ── Evaluate both models ──
lr_results  = evaluate_model('Logistic Regression', y_test, lr_pred,  lr_pred_proba[:, 1],
                              lr_train_time,  score_is_probability=True)
svm_results = evaluate_model('LinearSVC',            y_test, svm_pred, svm_scores,
                              svm_train_time, score_is_probability=False)

print('Evaluation complete for both models!')
print('  Logistic Regression AUC computed from : predict_proba()')
print('  LinearSVC AUC computed from            : decision_function() [uncalibrated scores]')
````

````python
# ── Display the comparison table ──
results_df = pd.DataFrame([lr_results, svm_results])
results_df = results_df.set_index('Model')

print('=' * 78)
print('   MODEL COMPARISON — FAKE NEWS DETECTION')
print('=' * 78)
print(results_df[['Accuracy', 'Precision', 'Recall', 'F1 Score', 'ROC-AUC', 'Training Time']].to_string())
print()
print('Note on ROC-AUC: Logistic Regression uses calibrated predict_proba() scores;')
print('LinearSVC uses uncalibrated decision_function() scores (valid for ranking/AUC only).')
print()

# Identify the best model by F1-Score (primary criterion)
best_model_name = results_df['F1 Score'].idxmax()
best_f1         = results_df['F1 Score'].max()
print('Best model by F1-Score:', best_model_name)
print('Best F1-Score         :', best_f1)
````

## Section 9: Confusion Matrices

A **Confusion Matrix** shows exactly where each model is making correct predictions and where it is making mistakes. Each cell tells us:

```
                 Predicted Fake   Predicted Real
Actual Fake  |   True Neg (TN)  |  False Pos (FP)  |
Actual Real  |   False Neg (FN) |  True Pos (TP)   |
```

- **Diagonal cells** (top-left, bottom-right) = correct predictions
- **Off-diagonal cells** = errors
- **FP (top-right)** = Fake news classified as Real — the most dangerous error
- **FN (bottom-left)** = Real news classified as Fake


````python
# ── Plot Confusion Matrices side by side ──
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

models_info = [
    ('Logistic Regression', lr_pred,  axes[0], '#3498db'),
    ('SVM (LinearSVC)',     svm_pred, axes[1], '#e74c3c'),
]

for model_name, preds, ax, color in models_info:
    cm = confusion_matrix(y_test, preds)
    sns.heatmap(
        cm,
        annot=True,
        fmt='d',
        cmap='Blues',
        ax=ax,
        linewidths=0.5,
        linecolor='gray',
        annot_kws={'size': 14, 'weight': 'bold'}
    )
    ax.set_title(model_name + '\nConfusion Matrix', fontsize=13, fontweight='bold', color=color)
    ax.set_xlabel('Predicted Label', fontsize=11)
    ax.set_ylabel('Actual Label', fontsize=11)
    ax.set_xticklabels(['Fake (0)', 'Real (1)'], fontsize=10)
    ax.set_yticklabels(['Fake (0)', 'Real (1)'], fontsize=10, rotation=0)

plt.suptitle('Confusion Matrices: Logistic Regression vs SVM', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('confusion_matrices.png', dpi=150, bbox_inches='tight')
plt.show()
print('Confusion matrices saved as confusion_matrices.png')
````

````python
# ── Print exact confusion matrix values with interpretation ──
for model_name, preds in [('Logistic Regression', lr_pred), ('SVM (LinearSVC)', svm_pred)]:
    cm = confusion_matrix(y_test, preds)
    tn, fp, fn, tp = cm.ravel()
    total = tn + fp + fn + tp
    print('=' * 52)
    print('  ' + model_name)
    print('=' * 52)
    print('  True  Negatives (Fake correctly caught) : {:5d}  ({:.1f}%)'.format(tn, tn/total*100))
    print('  False Positives (Fake called Real)      : {:5d}  ({:.1f}%)'.format(fp, fp/total*100))
    print('  False Negatives (Real called Fake)      : {:5d}  ({:.1f}%)'.format(fn, fn/total*100))
    print('  True  Positives (Real correctly found)  : {:5d}  ({:.1f}%)'.format(tp, tp/total*100))
    print('  Total test articles                     : {:5d}'.format(total))
    print()
````

## Section 11: Most Influential Features — What Does the Model Look For?

One of the strengths of Logistic Regression is **interpretability**. We can directly inspect the model's learned weights to see:
- Which words/bigrams most strongly predict **Real News** (high positive weight)
- Which words/bigrams most strongly predict **Fake News** (high negative weight)

This provides transparency and helps us understand what signals the model uses.


````python
# ── Extract feature importances from Logistic Regression ──
feature_names = tfidf_vectorizer.get_feature_names_out()
coef = lr_model.coef_[0]  # One coefficient per TF-IDF feature

# Top 25 features pointing toward REAL NEWS (highest positive coefficients)
top_real_idx  = np.argsort(coef)[::-1][:25]
top_real_words = [(feature_names[i], coef[i]) for i in top_real_idx]

# Top 25 features pointing toward FAKE NEWS (most negative coefficients)
top_fake_idx  = np.argsort(coef)[:25]
top_fake_words = [(feature_names[i], coef[i]) for i in top_fake_idx]

print('Top 25 features => REAL NEWS (high positive weights):')
print('{:5s}  {:35s}  {:s}'.format('Rank', 'Feature', 'Weight'))
print('-' * 55)
for rank, (word, w) in enumerate(top_real_words, 1):
    print('{:5d}  {:35s}  {:+.4f}'.format(rank, word, w))
````

````python
print('Top 25 features => FAKE NEWS (high negative weights):')
print('{:5s}  {:35s}  {:s}'.format('Rank', 'Feature', 'Weight'))
print('-' * 55)
for rank, (word, w) in enumerate(top_fake_words, 1):
    print('{:5d}  {:35s}  {:+.4f}'.format(rank, word, w))
````

## Section 12: Prediction Confidence Distribution

Let us examine how **confident** each model is in its predictions.

- For **Logistic Regression**, confidence = the max class probability from `predict_proba()` (a true probability between 0.5 and 1.0).
- For **LinearSVC**, there is no probability. Instead we look at the **absolute decision-function margin** (distance from the separating hyperplane). A larger absolute margin means the point sits farther from the boundary and the model is "more certain" in a geometric sense — but this is **not** on a 0–1 probability scale and the two are not directly comparable in magnitude. Calibrating LinearSVC into a true probability is left for the next task.


````python
# ── Confidence / margin distribution for both models ──
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Prediction Confidence Distribution', fontsize=14, fontweight='bold')

# --- Logistic Regression: true probability-based confidence ---
lr_confidence = lr_pred_proba.max(axis=1)
y_test_arr = np.array(y_test)

axes[0, 0].hist(lr_confidence, bins=40, color='#9b59b6', alpha=0.85, edgecolor='white')
axes[0, 0].set_title('Logistic Regression — Confidence Distribution (predict_proba)', fontsize=10)
axes[0, 0].set_xlabel('Max Predicted Probability')
axes[0, 0].set_ylabel('Number of Articles')
axes[0, 0].axvline(lr_confidence.mean(), color='red', linestyle='--',
                    label='Mean: {:.3f}'.format(lr_confidence.mean()))
axes[0, 0].legend()
axes[0, 0].grid(axis='y', alpha=0.3)

lr_correct_conf   = lr_confidence[lr_pred == y_test_arr]
lr_incorrect_conf = lr_confidence[lr_pred != y_test_arr]
axes[0, 1].hist(lr_correct_conf,   bins=30, alpha=0.7, color='#27ae60',
                 label='Correct ({})'.format(len(lr_correct_conf)), edgecolor='white')
axes[0, 1].hist(lr_incorrect_conf, bins=30, alpha=0.7, color='#e74c3c',
                 label='Wrong ({})'.format(len(lr_incorrect_conf)), edgecolor='white')
axes[0, 1].set_title('Logistic Regression — Correct vs Wrong Confidence', fontsize=10)
axes[0, 1].set_xlabel('Predicted Probability')
axes[0, 1].set_ylabel('Count')
axes[0, 1].legend(fontsize=9)
axes[0, 1].grid(axis='y', alpha=0.3)

# --- LinearSVC: uncalibrated decision-function margin (NOT a probability) ---
svm_margin = np.abs(svm_scores)

axes[1, 0].hist(svm_margin, bins=40, color='#f39c12', alpha=0.85, edgecolor='white')
axes[1, 0].set_title('LinearSVC — |Decision Margin| Distribution (NOT a probability)', fontsize=10)
axes[1, 0].set_xlabel('|decision_function score|')
axes[1, 0].set_ylabel('Number of Articles')
axes[1, 0].axvline(svm_margin.mean(), color='red', linestyle='--',
                    label='Mean: {:.3f}'.format(svm_margin.mean()))
axes[1, 0].legend()
axes[1, 0].grid(axis='y', alpha=0.3)

svm_correct_margin   = svm_margin[svm_pred == y_test_arr]
svm_incorrect_margin = svm_margin[svm_pred != y_test_arr]
axes[1, 1].hist(svm_correct_margin,   bins=30, alpha=0.7, color='#27ae60',
                 label='Correct ({})'.format(len(svm_correct_margin)), edgecolor='white')
axes[1, 1].hist(svm_incorrect_margin, bins=30, alpha=0.7, color='#e74c3c',
                 label='Wrong ({})'.format(len(svm_incorrect_margin)), edgecolor='white')
axes[1, 1].set_title('LinearSVC — Correct vs Wrong |Margin|', fontsize=10)
axes[1, 1].set_xlabel('|decision_function score|')
axes[1, 1].set_ylabel('Count')
axes[1, 1].legend(fontsize=9)
axes[1, 1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('confidence_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
print('Confidence/margin distribution saved as confidence_distribution.png')
print()
print('Reminder: LinearSVC margins are on a different, uncalibrated scale — do not')
print('compare their magnitude directly against Logistic Regression probabilities.')
````

## Section 13: Model Selection — Choosing the Best Model

We compare both models on all metrics and select the one that best balances **accuracy**, **F1-Score**, **ROC-AUC**, and **training efficiency**.


````python
# ── Full model comparison with winner indicators ──
print('=' * 78)
print('   FINAL MODEL COMPARISON — FAKE NEWS DETECTION')
print('=' * 78)
print()

metric_keys = ['Accuracy', 'Precision', 'Recall', 'F1 Score', 'ROC-AUC', 'Training Time']

print('{:18s}  {:22s}  {:22s}  {:s}'.format('Metric', 'Logistic Regression', 'LinearSVC', 'Winner'))
print('-' * 78)

for metric in metric_keys:
    lr_val  = lr_results[metric]
    svm_val = svm_results[metric]

    if metric == 'Training Time':
        # Lower is better for training time
        winner = 'Logistic Regression' if lr_val < svm_val else 'LinearSVC'
    else:
        winner = 'Logistic Regression' if lr_val > svm_val else ('LinearSVC' if svm_val > lr_val else 'TIE')

    print('{:18s}  {:22}  {:22}  {:s}'.format(metric, lr_val, svm_val, winner))

print()

# ── Select the best model — primarily by F1-Score, using the others as tie-breakers ──
lr_f1,  svm_f1  = lr_results['F1 Score'], svm_results['F1 Score']
lr_acc, svm_acc = lr_results['Accuracy'], svm_results['Accuracy']
lr_auc, svm_auc = lr_results['ROC-AUC'], svm_results['ROC-AUC']

reason_bits = []
if svm_f1 != lr_f1:
    winner_by_f1 = 'LinearSVC' if svm_f1 > lr_f1 else 'Logistic Regression'
    best_name = winner_by_f1
    reason_bits.append('higher F1-Score ({:.4f} vs {:.4f})'.format(max(svm_f1, lr_f1), min(svm_f1, lr_f1)))
else:
    # F1 tie -> break tie using Accuracy, then ROC-AUC
    if svm_acc != lr_acc:
        best_name = 'LinearSVC' if svm_acc > lr_acc else 'Logistic Regression'
        reason_bits.append('tied F1-Score, higher Accuracy')
    else:
        best_name = 'LinearSVC' if svm_auc >= lr_auc else 'Logistic Regression'
        reason_bits.append('tied F1-Score and Accuracy, higher/equal ROC-AUC')

if best_name == 'LinearSVC':
    best_model = svm_model
    best_res   = svm_results
else:
    best_model = lr_model
    best_res   = lr_results

selection_reason = 'Selected for ' + ', '.join(reason_bits) + '.'

print('=' * 78)
print()
print('SELECTED BEST MODEL :', best_name)
print('Accuracy             :', best_res['Accuracy'])
print('Precision             :', best_res['Precision'])
print('Recall                :', best_res['Recall'])
print('F1 Score              :', best_res['F1 Score'])
print('ROC-AUC               :', best_res['ROC-AUC'])
print('Training Time (s)     :', best_res['Training Time'])
print()
print('Reason for selection  :', selection_reason)
````

---

### Why did this model win?

The winner above is chosen **primarily by weighted evidence on F1-Score**, with Accuracy and ROC-AUC as tie-breakers — never a fixed assumption. Broadly:

**LinearSVC tends to do well because:**
- It is specifically designed to find the **maximum-margin** decision boundary
- It is robust to the high-dimensional, sparse nature of TF-IDF features
- It is one of the standard strong baselines for text classification

**Logistic Regression tends to do well because:**
- It is directly optimized for probabilistic classification and gives well-calibrated probabilities out of the box
- It is extremely interpretable (coefficients = word importance)
- On clean, linearly-separable-ish features like TF-IDF, it often matches SVM closely

> In practice the gap between the two is often small. The concrete numbers above (not an assumption) determine the actual winner for this run.
