## Section 3: The Reusable Prediction Function

`predict_news(text, title=None)` in `src/prediction.py`:

1. Validates the input (empty / whitespace-only / too short)
2. Preprocesses the text using the exact Task 2 logic
3. Transforms the cleaned text with the saved TF-IDF vectorizer (**transform only** — never re-fit)
4. Passes the vector to the saved model
5. Computes the prediction label using the confirmed label mapping
6. Computes a confidence score appropriate to the model type
7. Returns a structured dictionary

### Confidence handling (Section 5 of the spec)

- If the model has `predict_proba()` → confidence = the calibrated probability of the predicted class, and `confidence_type = "probability"`.
- If the model only has `decision_function()` (e.g. plain `LinearSVC`, as selected in Task 4) → there is **no calibrated probability available**. We report:
  - the raw signed `decision_score` (distance from the hyperplane), and
  - an `confidence` percentage obtained by squashing that score through a sigmoid **purely as a human-readable proxy**
  - `confidence_type = "decision_margin (uncalibrated)"` so this is never confused with a true probability.

We deliberately do **not** silently fit a calibrator (e.g. `CalibratedClassifierCV`) inside Step 5 — that would be retraining, which is out of scope here.

We test the pipeline with two realistic articles (Test 1, Test 2) and one invalid input (Test 3), exactly as required.

## Section 6: Batch Prediction

`predict_multiple()` runs `predict_news()` over a list of texts and returns a list of result dictionaries in the same order — useful for processing several articles at once (e.g. a CSV upload in the Streamlit app).

````
batch_texts = [test_article_1, test_article_2, "Too short"]
batch_results = pred.predict_multiple(batch_texts)

for i, r in enumerate(batch_results, 1):
    print(f'Article {i}:', r.get('prediction', r.get('error')),
          '' if 'error' in r else '({:.2f}% confidence)'.format(r['confidence']))
````

## Section 7: Compatibility Check (Step 2 → Step 3 → Step 4 → Step 5)

Before finishing, we confirm the whole chain is consistent:

- **Feature dimensions**: the vectorizer's vocabulary size must match what the model expects.
- **Vocabulary compatibility**: we only ever call `vectorizer.transform()`, never `fit()`, so the vocabulary learned in Task 3 is exactly preserved.
- **Label mapping**: confirmed directly from Task 1 (`Fake.csv` → 0, `True.csv` → 1), not assumed.
- **Model type**: read dynamically (`type(model).__name__`), not hardcoded.
- **Confidence calculation**: dynamically branches on `hasattr(model, 'predict_proba')` vs `hasattr(model, 'decision_function')`.

````
import scipy.sparse as sp

# Re-derive expected feature count directly from the vectorizer and confirm
# it matches what the loaded model expects.
vocab_size = len(pred.vectorizer.vocabulary_)

if hasattr(pred.model, 'coef_'):
    model_n_features = pred.model.coef_.shape[1]
elif hasattr(pred.model, 'n_features_in_'):
    model_n_features = pred.model.n_features_in_
else:
    model_n_features = None

print('Vectorizer vocabulary size :', vocab_size)
print('Model expected feature dim :', model_n_features)

if model_n_features is not None:
    assert vocab_size == model_n_features, 'MISMATCH: vectorizer/model feature dimensions differ!'
    print()
    print('Feature dimensions match — Step 3 vectorizer and Step 4 model are compatible.')

print()
print('Label mapping in use        :', pred.LABEL_NAMES)
print('Model classes_ (if present) :', getattr(pred.model, 'classes_', 'N/A'))
````

### Known, Documented Limitation

The current best model (`LinearSVC`) does not provide calibrated probabilities.
