## Section 4: Combine Title and Text Columns

News articles have two important text fields:
- **title** — a short headline summarizing the article
- **text** — the full body of the article

We combine them into a single column called `combined_text` so the model can learn from **both** the headline and the full article content.

> **Why combine?** The title often contains strong signals (e.g., clickbait language in fake news), and the body contains detailed content. Combining gives our model more information to work with.


```python
# Convert both columns to string first (safety measure in case of NaN values)
df['title'] = df['title'].astype(str)
df['text']  = df['text'].astype(str)

# Combine: title + space + text
# We add a space between them to prevent the last word of title merging with first word of text
df['combined_text'] = df['title'] + ' ' + df['text']

print('combined_text column created!')
avg_len = df['combined_text'].str.len().mean()
print(f'Average combined text length: {avg_len:.0f} characters')
print()
# Show an example
print('Example combined_text (first 300 characters):')
print('-' * 60)
print(df['combined_text'].iloc[0][:300])
```

## Section 5: Define the NLP Preprocessing Function

We define a single function called `preprocess_text()` that runs all preprocessing steps **in order**:

| Step | What it does | Why it helps |
|---|---|---|
| 1. Lowercase | Makes 'Apple' and 'apple' the same word | Reduces vocabulary size |
| 2. Remove HTML | Strips `<b>`, `<p>`, `<a>` tags | Some articles scraped from web retain HTML |
| 3. Remove URLs | Removes `http://...` links | URLs add no meaning |
| 4. Remove Emails | Removes user@example.com addresses | Noise reduction |
| 5. Remove Punctuation | Removes `.`, `,`, `!`, `?`, etc. | Models do not need them |
| 6. Remove Numbers | Removes standalone digits | Numbers rarely help classify fake vs real |
| 7. Tokenize | Splits sentence into individual words | Required for next steps |
| 8. Remove Stopwords | Removes 'the', 'is', 'at', 'which' etc. | Common words that carry no meaning |
| 9. Lemmatize | Converts words to their base form | 'running' to 'run', 'cars' to 'car' |
| 10. Rejoin | Joins tokens back into a string | For saving and model input |


```python
def preprocess_text(text):
    """
    Cleans and normalizes a raw news article text through 10 preprocessing steps.

    Args:
        text (str): Raw input text (combined title and article body)

    Returns:
        str: Cleaned, preprocessed text ready for machine learning
    """

    # Guard: handle NaN or non-string input gracefully
    if not isinstance(text, str):
        return ''

    # ── Step 1: Convert to lowercase ─────────────────────────────────────────
    # Ensures 'Trump' and 'trump' are treated as the same word.
    # Without this, the model sees them as two different tokens.
    text = text.lower()

    # ── Step 2: Remove HTML tags ──────────────────────────────────────────────
    # Some news articles are scraped from websites and retain HTML markup.
    # BeautifulSoup extracts only the visible text, stripping all HTML tags.
    # Example:  '<p>Hello <b>World</b></p>'  -->  'Hello World'
    text = BeautifulSoup(text, 'html.parser').get_text()

    # ── Step 3: Remove URLs ───────────────────────────────────────────────────
    # Removes web links like 'http://example.com' or 'https://www.bbc.co.uk'
    # URLs do not help classify articles as real or fake.
    text = re.sub(r'http\S+|www\.\S+', '', text)

    # ── Step 4: Remove email addresses ───────────────────────────────────────
    # Removes patterns like 'contact@newssite.com'
    text = re.sub(r'\S+@\S+', '', text)

    # ── Step 5: Remove punctuation and special characters ─────────────────────
    # Removes . , ! ? ; : ' " ( ) [ ] { } @ # $ % & etc.
    # We keep only word characters (letters, digits, underscore) and spaces.
    text = re.sub(r'[^\w\s]', '', text)

    # ── Step 6: Remove standalone numbers ────────────────────────────────────
    # Removes isolated digits like '2023', '42', '100'.
    # Note: Words like 'covid19' or 'mp3' are kept because they contain letters.
    text = re.sub(r'\b\d+\b', '', text)

    # ── Step 7: Remove extra whitespace ──────────────────────────────────────
    # After previous steps, there may be multiple spaces in a row.
    # This collapses them into a single space and removes leading/trailing whitespace.
    text = re.sub(r'\s+', ' ', text).strip()

    # ── Step 8: Tokenize ─────────────────────────────────────────────────────
    # Splits the text string into a Python list of individual words (called tokens).
    # Example:  'hello world news'  -->  ['hello', 'world', 'news']
    # Tokenization is required before we can apply stopword removal and lemmatization.
    tokens = word_tokenize(text)

    # ── Step 9: Remove stopwords ──────────────────────────────────────────────
    # Filters out very common English words that appear in almost every sentence
    # but carry very little meaning on their own.
    # Examples: 'the', 'a', 'is', 'at', 'which', 'on', 'and', 'or'
    # We also remove single-character tokens (single letters are usually noise).
    tokens = [word for word in tokens
              if word not in stop_words and len(word) > 1]

    # ── Step 10: Lemmatize ────────────────────────────────────────────────────
    # Reduces each word to its base (dictionary) form.
    # Examples:
    #   running  --> run
    #   better   --> good
    #   cars     --> car
    #   policies --> policy
    # This groups word variants together, reducing vocabulary size further.
    tokens = [lemmatizer.lemmatize(word) for word in tokens]

    # ── Step 11: Rejoin tokens into a single string ───────────────────────────
    # Convert the list of tokens back into a space-separated string.
    # This is the format needed for saving to CSV and for TF-IDF in Task 3.
    clean_text = ' '.join(tokens)

    return clean_text


print('preprocess_text() function defined successfully!')
print('The function is ready to process all 44,689 articles.')
```

## Section 6: Before and After Preprocessing Examples

Before running preprocessing on the full dataset, let us **test our function** on a few example articles and clearly see the transformation.


```python
# Show Before vs After for 3 sample articles from the actual dataset
sample_indices = [0, 1, 2]  # Use the first 3 articles

for i, idx in enumerate(sample_indices):
    raw   = df['combined_text'].iloc[idx]
    clean = preprocess_text(raw)
    label = 'REAL' if df['label'].iloc[idx] == 1 else 'FAKE'

    print('=' * 70)
    print(f'  EXAMPLE {i+1}  |  Label: {label} News  |  Row Index: {idx}')
    print('=' * 70)
    print(f'  BEFORE (first 400 chars):')
    print(f'  {raw[:400]}')
    print()
    print(f'  AFTER  (first 400 chars):')
    print(f'  {clean[:400]}')
    print()
    print(f'  Length: {len(raw)} chars  -->  {len(clean)} chars')
    print()
```

```python
# Edge case testing: verify each cleaning step handles special input correctly
edge_cases = [
    '<b>BREAKING NEWS:</b> Trump wins! Visit http://fakenews.com for more. 100 percent confirmed!!',
    'The president said on 21st January 2024, We will win. He also added: Its truly amazing.',
    'Scientists discovered a new planet in 2023. Running, ran, and runs are all forms of the same verb.',
    'URGENT SHARE THIS NOW everyone fakenews viral you wont BELIEVE this latest update'
]

edge_labels = [
    'HTML + URL + Numbers',
    'Dates + Capitalization',
    'Word variants (lemmatization test)',
    'Uppercase + noise words'
]

print('Edge Case Testing: Verifying Each Cleaning Step')
print()
for text, label in zip(edge_cases, edge_labels):
    cleaned = preprocess_text(text)
    print(f'  ---- {label} ----')
    print(f'  RAW  : {text}')
    print(f'  CLEAN: {cleaned}')
    print()
```

## Section 7: Apply Preprocessing to the Full Dataset

Now we apply the `preprocess_text()` function to all **44,689 articles** using pandas `.apply()` method.

> **Note:** Processing the full dataset with NLTK tokenization and lemmatization may take a few minutes depending on your machine. A progress bar from `tqdm` will show the progress.


```python
import time

# tqdm.pandas() enables a visual progress bar on the .apply() call
# This gives you feedback as each of the 44,689 rows is processed
tqdm.pandas(desc='Preprocessing articles')

rows_before = len(df)
print(f'Rows before preprocessing: {rows_before:,}')
print()

start_time = time.time()

# Apply the preprocessing function to each row in the 'combined_text' column.
# The cleaned result is stored in a new column called 'clean_text'.
# progress_apply() works like .apply() but shows a progress bar.
df['clean_text'] = df['combined_text'].progress_apply(preprocess_text)

elapsed = time.time() - start_time
print()
print(f'Preprocessing complete!')
print(f'Time taken: {elapsed:.1f} seconds ({elapsed/60:.1f} minutes)')
print(f'Rows after preprocessing: {len(df):,}')
```

## Section 8: Post-Preprocessing Quality Checks

After preprocessing, we must verify:
1. No missing values (`NaN`) in `clean_text`
2. No empty strings (articles that became empty after cleaning)
3. Row count matches (no rows were accidentally dropped)
4. Word count statistics (to understand how much text was reduced)


```python
# Check 1: Missing values in clean_text
# A missing value (NaN) would break the ML model in Task 3
missing_clean = df['clean_text'].isnull().sum()
print(f'Missing values in clean_text: {missing_clean}')

if missing_clean == 0:
    print('   No missing values found - clean_text is complete!')
else:
    print(f'   {missing_clean} missing values detected - filling with empty string.')
    df['clean_text'] = df['clean_text'].fillna('')
```

```python
# Check 2: Empty strings in clean_text
# Some articles might become completely empty after removing HTML, URLs, numbers, and stopwords.
# These would be articles that had no real textual content.
empty_rows = (df['clean_text'].str.strip() == '').sum()
print(f'Empty clean_text rows: {empty_rows}')

if empty_rows == 0:
    print('   No empty texts found - all articles have content after preprocessing!')
else:
    print(f'   {empty_rows} articles became empty after preprocessing.')
    print('   These may be articles that contained only noise (URLs, numbers, HTML).')
    print('   Showing up to 5 affected rows:')
    empty_df = df[df['clean_text'].str.strip() == ''][['title', 'combined_text', 'clean_text']]
    print(empty_df.head().to_string())
```

```python
# Check 3: Row count before vs after preprocessing
print(f'Rows BEFORE preprocessing: {rows_before:,}')
print(f'Rows AFTER  preprocessing: {len(df):,}')

if len(df) == rows_before:
    print('   Row count is unchanged - no rows were accidentally dropped!')
else:
    diff = rows_before - len(df)
    print(f'   WARNING: {diff} rows are missing. Please investigate.')
```

```python
# Check 4: Word count statistics before and after preprocessing
# This helps us understand how much the text was reduced
df['word_count_before'] = df['combined_text'].apply(lambda x: len(x.split()))
df['word_count_after']  = df['clean_text'].apply(lambda x: len(x.split()))

mean_before = df['word_count_before'].mean()
mean_after  = df['word_count_after'].mean()
med_before  = df['word_count_before'].median()
med_after   = df['word_count_after'].median()

reduction_pct = (1 - mean_after / mean_before) * 100

print('Word Count Statistics:')
print()
print(f'  Average words BEFORE : {mean_before:.1f}')
print(f'  Average words AFTER  : {mean_after:.1f}')
print()
print(f'  Median words BEFORE  : {med_before:.1f}')
print(f'  Median words AFTER   : {med_after:.1f}')
print()
print(f'  Word count reduced by approximately {reduction_pct:.1f}% after removing stopwords and noise.')
```
